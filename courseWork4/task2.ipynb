```python
import pandas as pd
import numpy as np
```

```python
def load_dataset():
    SportCar = pd.read_csv("D:\Python Course work\courseWork4\Sport car price.csv")
    return SportCar
```

```python
def clean_dataset(df):
    """
    Remove missing values and duplicate rows.
    Convert appropriate columns to numeric values.
    """
    cleaned_df = df.copy()

    # Remove duplicate rows
    cleaned_df = cleaned_df.drop_duplicates()

    # Columns that should contain numeric data
    numeric_columns = [
        "Year",
        "Engine Size (L)",
        "Horsepower",
        "Torque (lb-ft)",
        "0-60 MPH Time (seconds)",
        "Price (in USD)"
    ]

    # Convert columns to numeric
    for column in numeric_columns:
        cleaned_df[column] = pd.to_numeric(
            cleaned_df[column],
            errors="coerce"
        )

    # Remove rows containing missing values
    cleaned_df = cleaned_df.dropna()

    return cleaned_df
```
